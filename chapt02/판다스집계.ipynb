## 판다스 집계

### 집계
> 계산, 통계, 그룹핑으로 원데이터에서 수치적으로 필요한 값을 재가공

```python
import numpy as np
import pandas as pd
```

```python
orders = pd.DataFrame({
    "order_id": [1, 2, 3, 4, 5, 6],
    "customer": ["홍길동", "김철수", "이영희", "홍길동", "박민수", "김철수"],
    "product_id": ["P001", "P002", "P001", "P003", "P002", "P003"],
    "quantity": [2, 1, 3, 1, 2, 4],
    "unit_price": [30000, 15000, 30000, 8000, 15000, 8000],
    "order_date": [
        "2026-09-01",
        "2026-09-01",
        "2026-09-02",
        "2026-09-02",
        "2026-09-03",
        "2026-09-03"
    ]
})

# 주문번호, 고객명, 상품ID, 수량, 단가, 주문일자로 구성된 데이터
orders
```

#### 파생 컬럼 생성

- 단가 * 수량 > 주문금액 컬럼 생성

```python
orders['total_price'] = orders['quantity'] * orders['unit_price']
```

```python
orders
```

#### 조건을 이용한 컬럼 생성

- 조건 연산자 (> , < ,>= , <=)와 논리연산자(&, |, ~) 등을 사용해서 필요한 조건에 맞는 데이터만 필터링
- 필터링 후 파생컬럼 생성

```python
orders['customer_type'] = '일반'

orders
```

```python
orders['total_price'] > 50000 
```

```python
orders.loc[condition]
```

```python
orders.loc[orders['total_price'] >= 50000]
```

```python
orders.loc[orders['total_price'] >= 50000, 'customer_type'] = 'VIP'
```

```python
orders
```

#### apply() 함수 사용
- apply() 함수 내에 사용할 사용자 정의 함수
- 학점 등 여러 구분으로 나눠지는 도메인 값의 컬럼을 생성할때 유용

```python
def get_order_grade(total_price):
    if total_price >= 80000:
        return 'A'
    elif total_price >= 30000:
        return 'B'
    else:
        return 'C'
```

```python
get_order_grade(85000)
```

```python
get_order_grade(25000)
```

```python
# total_price 컬럼의 값을 사용해서 get_order_grade 함수를 적용하여 행별로 grade 컬럼을 생성
orders['grade'] = orders['total_price'].apply(get_order_grade)
```

```python
orders['order_date'] = pd.to_datetime(orders['order_date'])
```

```python
# 날짜형식으로 변경하면, 년.월.일 등으로 분리해서 데이터 사용이 가능
orders['order_date'].dt.day
```

#### CONCAT
- 데이터 합치기

```python
week1 = pd.DataFrame({
    "order_id": [1, 2],
    "customer": ["홍길동", "김철수"],
    "amount": [60000, 15000]
})
 
week2 = pd.DataFrame({
    "order_id": [3, 4],
    "customer": ["이영희", "박민수"],
    "amount": [90000, 30000]
})
```

```python
week1
```

```python
week2
```

```python
# 이전 인덱스가 그대로 반영
pd.concat([week1, week2])
```

```python
# 일반적으로 가장 많이 사용하는 방법
pd.concat([week1, week2], ignore_index=True)
```

#### Merge
- DB Join과 동일

```python
products = pd.DataFrame({
    "product_id": ["P001", "P002", "P003"],
    "product_name": ["키보드", "마우스", "USB"],
    "category": ["주변기기", "주변기기", "저장장치"]
})
```

```python
orders
```

```python
products
```

```python
pd.merge(orders, products, on='product_id')
```

- how : inner, left, right 가 99% 사용

```python
pd.merge(orders, products, on='product_id', how='left')
```

```python
products.loc[3] = ['P004', '모니터', '디스플레이']
```

```python
products
```

```python
pd.merge(orders, products, on='product_id', how='left')
```

```python
pd.merge(orders, products, on='product_id', how='right')
```

#### group by 집계
- DB group by 동일

```python
orders.groupby('customer')
```

```python
orders.groupby('customer')['total_price'].sum()
```

```python
# 데이터프레임으로 변환
orders.groupby('customer')['total_price'].sum().reset_index()
```

#### 여러 통계를 한번에 계산

```python
# 고객별로 기본통계
orders.groupby('customer')['total_price'].agg(
    ['count', 'sum', 'mean', 'max', 'min','std']
)
```

```python
# 통계치 csv 파일로 저장
orders.groupby('customer')['total_price'].agg(
    ['count', 'sum', 'mean', 'max', 'min','std']
).to_csv('../data/고객별구매통계.csv', encoding='utf-8')
```
