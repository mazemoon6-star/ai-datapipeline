## 데이터 수집 자동화

### 웹구조

#### Live Server 설치
> 간단한 웹페이지와 css를 확인 할 수 있는 웹서버 실행 플러그인
- VS Code 확장 > Live Server 검색 후 설치

#### 웹 페이지
- HTML, CSS , Javascript로 구성
    - `HTML`: 웹 페이지 구성, 구조, 내용
    - `CSS` : 글자, 색상, 배치 등의 디자인 파일
    - `Javascript` : 클릭, 입력, 화면 변경, 애니메이션 등 동작 

#### 데이터 수집 순서
1. HTML 구조 확인
2. 필요한 요소 찾기
3. 텍스트/이미지 추출
4. DataFrame 저장/이미지 파일로 저장

##### HTML 주요태그

| 태그                                                     | 의미                                 |
| -------------------------------------------------------- | ------------------------------------ |
| `<html>`                                               | HTML 문서 전체                       |
| `<head>`                                               | 문서 정보(화면에 표현x)              |
| `<body>`                                               | 문서 페이지(화면에 보이는 내용)      |
| `<title>`                                              | 웹 브라우저 탭 제목                  |
| `<h1>~<<h6>`                                           | 제목 1 ~ 6까지                       |
| `<p>`                                                  | 문단                                 |
| `<a>`                                                  | 링크/ 페이지 이동                    |
| `<img>`                                                | 이미지                               |
| `<ul>`,`<li>`                                        | 목록, 번호목록                       |
| `<table>`                                              | 표, 테이블                           |
| `<input>`                                              | 입력창, 텍스트 입력창, 콤보박스, ... |
| `<button>`                                             | 버튼                                 |
| `<div>`, `<span>`, `<section>`, `<content>`, ... | 영역구분                             |

##### 주요 속성
>  HTML 태그에는 여러 속성이 붙을 수 있음

```html
<div class = "product", id="item-1">
  <span class="name">키보드</span>
  <span class="price">20000</span>
</div>
```

|속성|의미|
|---|---|
|`id`| 페이지 내에서 중복되지 않는 값(id는 중복되면 안되나 html에서는 오류는 생성하지않음)|
|`class`| CSS 디자인을 위해서 사용. 여러 요소에 공통으로 붙이는 이름. 중복됨|
|`href` | 링크주소|
|`src`| 이미지나 소스 위치|
|`name`| 입력 요소 이름. 중복됨|

##### 개발자 도구
>웹 페이지에서 원하는 요소를 찾으려면 브라우저 개발자 도구를 사용. F12로 동작
- 웹 페이지 우클릭 > 검사
- 요소 탭에서 필요 요소 찾기
- HTML 구조 확인
- 스크랩핑 할 데이터 위치 찾기
- 태그 이름, id, class 확인
