- Series 작성시 인덱스를 직접 커스터마이징 할 수 있음

```python
scores = pd.Series(
    [85,90,95,80],
    index=['홍길동','이영희','김철수','박민수']
    )

scores
```

```python
# 기존 인덱스는 사용불가. 인덱스를 커스터마이징했기때문
scores[0]
```

```python
# 인덱스가 숫자에서 문자열로 변경
scores['홍길동']
```

#### DataFrame
> Pandas에서 2차원 표형태 데이터를 표현하는 자료 구조
- Pandas의 기본 자료구조형
- Excel, CSV(excel형태 텍스트파일), DB 데이블과 거의 동일한 구조

```python
# DataFrame을 직접 생성할때 사용.
# 파이썬 딕셔너리 사용
raw_data = {
    'name' : ['홍길동','김철수','이영희','박민수','성유고','애슐리'],
    'age' : [20,21,22,23,30,29],
    'major' : ['데이터분석', '웹개발', '인공지능','임베디드','인공지능','웹개발'],
    'score' : [85,90,95,80,90,100]
}
raw_data
```

```python
# 일반적으로 DataFrame을 줄여서 df로 변수명 많이 활용
df = pd.DataFrame(raw_data)

df
```
