### Python 리스트(배열)과 Numpy 배열 비교
- 파이썬에 기존 리스트(배열) 자료구조가 존재하나 속도적인 면에서 Numpy가 우세함

### 파이썬으로 배열작업

````python
prices = [10000, 15000, 12000, 30000, 25000] # 5개 가격이 든 배열
discount_prices = []

for price in prices:
    discount_prices.append(price * 0.9) # 10%할인

discount_prices # 노트북 코드쉘에서 하나만 출력하려면 print() 생략 가능
````

#### 넘파이로 배열작업

````python
np_prices =np.array(prices) # 파이썬 배열을 넘파이 배열로 변경
np_discount_prices = np_prices * 0.9

np_discount_prices
````

#### Numpy 데이터타입
- int8 : -128 ~ 127 (총 256)
- int16 : -32768 ~ 32676 (총 65536)
- int32 : 32비트 정수
- int64 : 64비트 정수 (Numpy 기본 정수)
- uint8 : 0 ~ 255
- unit16 : 0 ~ 65535
- float16 : 실수
- float32 : 32비트 실수
- float64 : 64비트 실수(Numpy 기본 실수)
- complex64 : 복소수
- complox128 : 128비트 복소수
- bool : True/False
- str_ : 문자열
- object_ : 객체
