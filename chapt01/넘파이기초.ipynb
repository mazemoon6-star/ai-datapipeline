````python
prices = [10000, 15000, 12000, 30000, 25000] # 5개 가격이 든 배열
discount_prices = []

for price in prices:
    discount_prices.append(price * 0.9) # 10%할인

discount_prices # 노트북 코드쉘에서 하나만 출력하려면 print() 생략 가능
````

- 결측치를 정리하는 방법 : 결측치는 제거하거나(건수가 줄어듬), 나머지 평균값을 임의로 기입하거나, 기본통계에서 도출할 수 있는 다른 값들로 채우기

### 이상치 핸들링
- 일반적인 범위를 벗어나는 비정상 데이터
예. 부산 1월 일별 평균기온 데이터 중 -90도, 습도 99% 

````python
temperatures = np.array([22.5, 23.0, np.nan, 25.1, 999.0, 21.8, -99.0])
temperatures
````

````python
valid_temperatures = temperatures[
    (~np.isnan(temperatures)) & # nan 무조건 제거
     (temperatures >= 15.0) &   # 최저기온이 15도 아래로 내려갈수없다 (제거)
     (temperatures <= 50)       # 최고기온이 50도 이상 올라갈수없다 (제거)
]

valid_temperatures
````

- 전처리(PreProcessing) : 분석 이전에 이상치, 결측치를 제거하는 작업

#### 매출 데이터 전처리
- 결측값 출력
- 결측값 제외
- 0원 ~ 10만원 범위의 정상 매출만 남기고 
- 정상 매출의 평균, 최대/최소값 출력
- 정상매출에서 평균이상인 값만 출력

````python
# 파이썬 리스트에서 None 대신 np.nan 으로 결측치를 입력하면 Numpy 패키지 연산이 가능

raw_sales = [12000,15000,np.nan,30000,-5000,25000, 99999, 18000, np.nan, 22000]
raw_sales
````

````python
# Numpay 배열로 변경
np_raw_sales = np.array(raw_sales, dtype=float)

np_raw_sales
````

````python
np.isnan(np_raw_sales)
````

````python
np.isnan(np_raw_sales).sum()
````

````python
valid_sales = np_raw_sales[
    (~np.isnan(np_raw_sales)) & 
    (np_raw_sales >= 0) &
    (np_raw_sales <= 100000)
]

valid_sales
````

````python
valid_sales.mean()
````

````python
valid_sales.min()
````

````python
valid_sales.max()
````

````python
mean_sales = valid_sales.mean()
mean_sales
````

````python
# 평균 매출액보다 높은 매출 찾기
valid_sales[valid_sales > mean_sales]
````
